```python
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter

# Read processed data
df = pd.read_csv('data/processed/processed.csv')
print('data shape:', df.shape)
df.head()
```

```python
# Class distribution plot
import os

counts = df['label'].value_counts().sort_index()
fig = plt.figure(figsize=(6,4))
counts.plot(kind='bar')
plt.title('Class distribution')
plt.xlabel('label')
plt.ylabel('count')
plt.tight_layout()

plt.show()
```

```python
# Message length distribution and top tokens
df['length'] = df['text'].astype(str).str.len()
plt.figure(figsize=(6,4))
df['length'].hist(bins=50)
plt.title('Message length distribution')
plt.xlabel('length')
plt.ylabel('count')
plt.tight_layout()

plt.show()

# Top tokens (TF-IDF, top features)
from sklearn.feature_extraction.text import TfidfVectorizer
vec = TfidfVectorizer(stop_words='english', max_features=20)
X = vec.fit_transform(df['text'].astype(str))
top = vec.get_feature_names_out()
print('Top tokens:', top)
```
